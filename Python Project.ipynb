```python
import json
import os

filename = "students.json"

# Load Data
if os.path.exists(filename):
    with open(filename, "r") as file:
        students = json.load(file)
else:
    students = {}

# Save Data
def save_data():
    with open(filename, "w") as file:
        json.dump(students, file)

# Add Student
def add_student():
    roll = input("Enter Roll Number: ")

    if roll in students:
        print("Student already exists!")
        return

    name = input("Enter Name: ")

    try:
        marks = float(input("Enter Marks: "))
    except ValueError:
        print("Invalid Marks!")
        return

    students[roll] = {
        "Name": name,
        "Marks": marks
    }

    save_data()
    print("\nStudent Added Successfully!\n")

    # Show all students after adding
    view_students()

# View Students
def view_students():
    if not students:
        print("No Student Records Found.")
        return

    print("\nStudent Details")
    print("-" * 40)

    for roll, details in students.items():
        print("Roll Number :", roll)
        print("Name        :", details["Name"])
        print("Marks       :", details["Marks"])
        print("-" * 40)

# Search Student
def search_student():
    roll = input("Enter Roll Number: ")

    if roll in students:
        print("Student Found")
        print("Roll Number :", roll)
        print("Name        :", students[roll]["Name"])
        print("Marks       :", students[roll]["Marks"])
    else:
        print("Student Not Found.")

# Update Marks
def update_marks():
    roll = input("Enter Roll Number: ")

    if roll in students:
        try:
            marks = float(input("Enter New Marks: "))
            students[roll]["Marks"] = marks
            save_data()
            print("Marks Updated Successfully!")
        except ValueError:
            print("Invalid Marks!")
    else:
        print("Student Not Found.")

# Delete Student
def delete_student():
    roll = input("Enter Roll Number: ")

    if roll in students:
        del students[roll]
        save_data()
        print("Student Deleted Successfully!")
    else:
        print("Student Not Found.")

# Main Menu
while True:
    print("\n===== Student Management System =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Marks")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        view_students()

    elif choice == "3":
        search_student()

    elif choice == "4":
        update_marks()

    elif choice == "5":
        delete_student()

    elif choice == "6":
        print("Thank You!")
        break

    else:
        print("Invalid Choice! Please Try Again.")
```

```python
# Load Questions
def load_questions():
    questions = []

    try:
        with open("question.txt", "r") as file:
            for line in file:
                data = line.strip().split("|")

                question = {
                    "question": data[0],
                    "a": data[1],
                    "b": data[2],
                    "c": data[3],
                    "answer": data[4]
                }

                questions.append(question)

    except FileNotFoundError:
        print("Question file not found!")

    return questions


# Conduct Quiz
def start_quiz():

    questions = load_questions()

    if len(questions) == 0:
        return

    score = 0

    print("\n===== QUIZ STARTED =====\n")

    for q in questions:

        print(q["question"])
        print(q["a"])
        print(q["b"])
        print(q["c"])

        ans = input("Enter Your Answer (a/b/c): ").lower()

        if ans == q["answer"]:
            print("Correct Answer!\n")
            score += 1
        else:
            print("Wrong Answer!")
            print("Correct Answer is:", q["answer"])
            print()

    print("Quiz Completed!")
    print("Your Score:", score, "/", len(questions))

    # Save Score
    with open("score.txt", "a") as file:
        name = input("Enter Your Name: ")
        file.write(f"{name} : {score}/{len(questions)}\n")

    print("Score Saved Successfully!")


# Main Program
start_quiz()
```

```python
import json
import os

class Library:

    def __init__(self):
        self.filename = "books.json"

        if os.path.exists(self.filename):
            with open(self.filename, "r") as file:
                self.books = json.load(file)
        else:
            self.books = {}

    # Save Data
    def save_data(self):
        with open(self.filename, "w") as file:
            json.dump(self.books, file)

    # Add Book
    def add_book(self):
        book_id = input("Enter Book ID: ")

        if book_id in self.books:
            print("Book already exists!")
            return

        title = input("Enter Book Name: ")
        author = input("Enter Author Name: ")

        self.books[book_id] = {
            "Title": title,
            "Author": author,
            "Status": "Available"
        }

        self.save_data()
        print("Book Added Successfully!")

    # Issue Book
    def issue_book(self):
        book_id = input("Enter Book ID: ")

        if book_id in self.books:

            if self.books[book_id]["Status"] == "Available":
                self.books[book_id]["Status"] = "Issued"
                self.save_data()
                print("Book Issued Successfully!")
            else:
                print("Book is already issued.")

        else:
            print("Book Not Found!")

    # Return Book
    def return_book(self):
        book_id = input("Enter Book ID: ")

        if book_id in self.books:

            if self.books[book_id]["Status"] == "Issued":
                self.books[book_id]["Status"] = "Available"
                self.save_data()
                print("Book Returned Successfully!")
            else:
                print("Book is already available.")

        else:
            print("Book Not Found!")

    # View Books
    def view_books(self):

        if not self.books:
            print("No Books Available.")
            return

        print("\n========== BOOK LIST ==========")

        for book_id, details in self.books.items():

            print("--------------------------------")
            print("Book ID :", book_id)
            print("Title   :", details["Title"])
            print("Author  :", details["Author"])
            print("Status  :", details["Status"])

        print("--------------------------------")


# Main Program
library = Library()

while True:

    print("\n====== Library Management System ======")
    print("1. Add Book")
    print("2. Issue Book")
    print("3. Return Book")
    print("4. View Books")
    print("5. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        library.add_book()

    elif choice == "2":
        library.issue_book()

    elif choice == "3":
        library.return_book()

    elif choice == "4":
        library.view_books()

    elif choice == "5":
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```

```python
import json
import os

filename = "expenses.json"

# Load Expenses
if os.path.exists(filename):
    with open(filename, "r") as file:
        expenses = json.load(file)
else:
    expenses = []


# Save Expenses
def save_data():
    with open(filename, "w") as file:
        json.dump(expenses, file)


# Add Expense
def add_expense():

    date = input("Enter Date (DD-MM-YYYY): ")
    category = input("Enter Category: ")

    try:
        amount = float(input("Enter Amount: "))
    except ValueError:
        print("Invalid Amount!")
        return

    expense = {
        "Date": date,
        "Category": category,
        "Amount": amount
    }

    expenses.append(expense)
    save_data()

    print("Expense Added Successfully!\n")

    # Show all expenses after adding
    view_expenses()


# View Expenses
def view_expenses():

    if len(expenses) == 0:
        print("No Expenses Found.")
        return

    print("\n========== Expense List ==========")

    for expense in expenses:

        print("----------------------------------")
        print("Date     :", expense["Date"])
        print("Category :", expense["Category"])
        print("Amount   : ₹", expense["Amount"])

    print("----------------------------------")


# Calculate Total Expense
def total_expense():

    total = 0

    for expense in expenses:
        total += expense["Amount"]

    print("\nTotal Expense : ₹", total)


# Main Menu
while True:

    print("\n====== Expense Tracker ======")
    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Calculate Total Expense")
    print("4. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        add_expense()

    elif choice == "2":
        view_expenses()

    elif choice == "3":
        total_expense()

    elif choice == "4":
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```
